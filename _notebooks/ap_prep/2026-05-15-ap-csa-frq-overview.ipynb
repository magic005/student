## FRQ Workflow



<pre class="mermaid">
flowchart LR
    A[Read class contract] --> B[Identify fields and helper methods]
    B --> C[Write method skeleton]
    C --> D[Trace sample cases]
    D --> E[Fix syntax and edge cases]
    E --> F[Compare against scoring guideline ideas]
</pre>


This workflow is why FRQ practice also helped MCQ: the same skeleton-first thinking turns a confusing answer choice into a traceable piece of Java.

## Repeating Trends

| Pattern | Where it showed up | Fix I practiced |
|---|---|---|
| Missing or misplaced return values | SignedText, WordChecker, WeatherData | State the method return type before writing branches. |
| Loop bounds | AppointmentBook, 2D grids, ArrayLists | Prefer `< length` and test first/last index mentally. |
| Mutating collections safely | WeatherData, ArrayList drills | Remove from the end when deleting during traversal. |
| String APIs | SignedText, Sign, WordChecker | Verify `substring`, `startsWith`, `endsWith`, and `contains` signatures. |
| 2D traversal | SumOrSameGame, BoxOfCandy, Data | Write row and column loops separately before adding logic. |
