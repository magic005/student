## Pattern I Saw

This set exposed a wider spread: arithmetic tracing, boolean equivalence, loop conditions, 1D arrays, random number ranges, and 2D arrays. The big trend was that the wrong answers mostly came from moving too quickly through code that mutates a variable or index.

```java
public class SetBRepairs {
    public static int randomInclusive(int low, int high) {
        return (int) (Math.random() * (high - low + 1)) + low;
    }

    public static int countNegatives(int[] nums) {
        int count = 0;
        for (int i = 0; i < nums.length; i++) {
            if (nums[i] < 0) {
                count++;
            }
        }
        return count;
    }

    public static int sum2D(int[][] nums) {
        int sum = 0;
        for (int r = 0; r < nums.length; r++) {
            for (int c = 0; c < nums[0].length; c++) {
                sum += nums[r][c];
            }
        }
        return sum;
    }

    public static boolean sameDecision(int x, int y) {
        boolean original = !(x < 0 || y < 0);
        boolean simplified = x >= 0 && y >= 0;
        return original == simplified;
    }
}
```
