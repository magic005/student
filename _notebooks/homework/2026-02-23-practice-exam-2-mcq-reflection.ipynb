## Detailed Analysis of Incorrect Answers

### Q10: Value of d after operations

**Topic:** 1.3 (Expressions)  
**Time Spent:** 2:02  
**My Answer:** B (18.0)  
**Correct Answer:** C (20.0)  

#### The Problem
```java
int a = 3 + 2 * 3;
int b = 4 + 3 / 2;
int c = 7 % 4 + 3;
double x = a + b + c;
```

#### Why I Was Wrong
I messed up `b`. I evaluated it as `(4 + 3) / 2 = 3`, but Java respects operator precedence: division happens before addition. So:
- `a = 3 + (2 * 3) = 9`
- `b = 4 + (3 / 2) = 4 + 1 = 5` (integer division!)
- `c = (7 % 4) + 3 = 3 + 3 = 6`
- `x = 9 + 5 + 6 = 20.0`

#### Key Takeaway
Always evaluate `*`, `/`, and `%` before `+` and `-`. I knew this rule but didn't apply it carefully for `b`. Integer division truncation (`3/2 = 1`) is also easy to forget under time pressure.

### Q12: Expression equivalent to nested ifs

**Topic:** 2.4, 2.5 (Selection, Boolean Logic)  
**Time Spent:** 2:12  
**My Answer:** C  
**Correct Answer:** D  

#### The Problem
Translate this nested if structure into a single boolean:
```java
boolean b1 = true;
if (num > 0) {
   if (num >= 100) { b1 = false; }
} else {
   if (num >= -100) { b1 = false; }
}
```

#### Why I Was Wrong
I chose `(num < -100) && (num > 0 || num < 100)`, which gives a wrong result for values between 0 and 100 exclusive. Let me trace the original logic:

- `b1` stays `true` when: `num > 0 AND num < 100` (positive but under 100), **OR** `num < -100` (very negative)
- `b1` becomes `false` when: `num >= 100`, OR `num <= 0 AND num >= -100`

The correct equivalent is:
```java
boolean b2 = (num < -100) || (num > 0 && num < 100);
```
This captures both cases where `b1` stays `true`: either the number is below -100, or it's strictly between 0 and 100.

#### Key Takeaway
When converting nested ifs to boolean expressions, I should trace several test values through the original code to build a truth table. I tried to reason abstractly and made a logic error. Concrete values like `num = 50` or `num = -50` would have caught this.

### Q17: Methods returning same value

**Topic:** 1.3, 2.8 (Expressions, Iteration)  
**Time Spent:** 1:24  
**My Answer:** D (when a % b == 1)  
**Correct Answer:** C (when a % b == 0)  

#### The Problem
Two methods, `methodX` (uses `a / b` in a for-loop bound) and `methodY` (increments by `b` in a while-loop until reaching `a`), and the question asks when they return the same value.

#### Why I Was Wrong
`methodX` returns `a / b` (integer division, which truncates). `methodY` returns the ceiling of `a / b` (since it counts how many times you can add `b` before exceeding `a`). These are equal only when there's no remainder, i.e., `a % b == 0`.

Example: `a = 7, b = 3`
- `methodX`: loop runs `7/3 = 2` times → returns 2
- `methodY`: `i=0→3→6→9`, counts 3 iterations → returns 3
- They differ because 7 % 3 = 1 ≠ 0

Example: `a = 6, b = 3`
- `methodX`: `6/3 = 2` → returns 2
- `methodY`: `i=0→3→6`, counts 2 → returns 2
- They match because 6 % 3 = 0

#### Key Takeaway
Integer division truncates in Java. A while-loop that increments by `b` effectively computes the ceiling. They only agree when division is exact. I need to be more careful about the distinction between floor and ceiling division.

Which code segment correctly counts negative values in an `int[] nums`?

### Q27: mystery array traversal

**Topic:** 4.4 (Array Traversal)  
**Time Spent:** 1:44  
**My Answer:** D (17)  
**Correct Answer:** B (7)  

#### The Problem
```java
public static int mystery(int[] arr) {
   int x = 0;
   for (int k = 0; k < arr.length; k = k + 2) {
      x = x + arr[k];
   }
   return x;
}
int[] nums = {3, 6, 1, 0, 1, 4, 2};
```

#### Why I Was Wrong
I summed ALL elements (3+6+1+0+1+4+2 = 17) instead of only the even-indexed ones. The loop increments by 2, so it only hits indices 0, 2, 4, 6:
- `arr[0] = 3`
- `arr[2] = 1`
- `arr[4] = 1`
- `arr[6] = 2`
- Sum = 3 + 1 + 1 + 2 = **7**

#### Key Takeaway
I didn't read the loop increment carefully. `k = k + 2` is not the standard `k++`. This is another classic AP trap. I need to slow down and trace the actual index values.

### Q33: Generate random number

**Topic:** 1.11 (Math Class)  
**Time Spent:** 1:13  
**My Answer:** C (6, 7, 8, or 9)  
**Correct Answer:** A (5, 6, 7, 8, or 9)  

#### The Problem
```java
int b = 5 + (int) (Math.random() * 5);
```

#### Why I Was Wrong
I thought `(int)(Math.random() * 5)` produces 1-4. It actually produces **0-4**:
- `Math.random()` → [0.0, 1.0)
- `* 5` → [0.0, 5.0)
- `(int)` truncates → 0, 1, 2, 3, or 4
- `+ 5` → 5, 6, 7, 8, or 9

The minimum is 5 (when `Math.random()` is near 0), not 6. I forgot that `Math.random()` includes 0.0.

#### Key Takeaway
`Math.random()` starts at 0.0 inclusive. The formula `(int)(Math.random() * n) + start` gives values from `start` to `start + n - 1`. For this problem: `5` to `5 + 5 - 1 = 9`.

### Q35: Add up numbers in 2D array

**Topic:** 4.13 (2D Array Traversal)  
**Time Spent:** 2:48  
**My Answer:** C (21)  
**Correct Answer:** D (26)  

#### The Problem
```java
int[][] values = { {1, 2, 3}, {4, 5, 6} };
int x = 0;
for (int j = 0; j < values.length; j++) {
   for (int k = 0; k < values[0].length; k++) {
      if (k == 0) {
         values[j][k] *= 2;
      }
      x += values[j][k];
   }
}
```

#### Why I Was Wrong
I added up 1+2+3+4+5+6 = 21 and ignored the `*= 2` on the first column. The code doubles `values[j][0]` before adding it to `x`:
- Row 0: `values[0][0]` becomes `1*2=2`, then adds 2+2+3 = 7
- Row 1: `values[1][0]` becomes `4*2=8`, then adds 8+5+6 = 19
- Total: 7 + 19 = **26**

#### Key Takeaway
I skipped the mutation inside the loop. When tracing 2D array problems, I need to track modifications to the array itself, not just the accumulator variable. The `*= 2` happens before the `+=`, so the doubled value is what gets added.

4. **Boolean expression translation** — When converting nested ifs to compound expressions:
   - Build a truth table with at least 5 test values
   - Check boundary values (0, -100, 100 in the Q12 case)
   - Verify every answer choice against the truth table
