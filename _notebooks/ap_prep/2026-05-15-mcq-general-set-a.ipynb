## What Improved

The misses were not syntax-heavy; they were reasoning misses. The `minOfThree` item showed that nested branches can accidentally skip a comparison. The recursive binary search item showed that I needed to trace `low`, `high`, and `mid` rather than trusting the shape of the algorithm.

```java
public class SetARepairs {
    public static int minOfThree(int a, int b, int c) {
        if (a <= b && a <= c) {
            return a;
        } else if (b <= a && b <= c) {
            return b;
        }
        return c;
    }

    public static int binarySearch(int[] values, int target) {
        return binarySearch(values, 0, values.length - 1, target);
    }

    private static int binarySearch(int[] values, int low, int high, int target) {
        if (low > high) {
            return -1;
        }
        int mid = (low + high) / 2;
        if (values[mid] == target) {
            return mid;
        } else if (target < values[mid]) {
            return binarySearch(values, low, mid - 1, target);
        }
        return binarySearch(values, mid + 1, high, target);
    }
}
```

## Takeaway

After this set, my rule was to trace every recursive call and every compound boolean condition. If I could not say why a branch was unreachable or reachable, I was guessing.
