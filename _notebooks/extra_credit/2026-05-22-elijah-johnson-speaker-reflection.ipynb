## Before the Event

Before attending, I expected a normal guest speaker talk about careers and college advice. I thought it would be useful, but I did not expect it to feel as personal or motivating as it did.

I was curious about how someone builds momentum in computer science outside of just doing assignments. I wanted to know what actually matters when you are trying to grow from a student into someone who can contribute to real technical work.

## What I Saw

During the event, I saw Elijah Johnson present to our class, answer questions, and turn the talk into a real conversation instead of just a lecture. I participated a lot by answering questions, asking my own questions, and staying engaged while he talked through his experience.

The most interesting thing I saw was how naturally he connected technical growth with initiative. I looked him up afterward and found an Elijah Johnson profile connected with EECS, hardware/software systems, and edge devices, which matched the way the talk made computer science feel practical instead of abstract.

## What I Learned

One specific thing I learned is that career growth in CS is not only about knowing syntax or finishing projects. It is also about asking better questions, following up with people, and being able to explain what you are interested in.

What surprised me was how much the interaction mattered. Asking questions during the talk and speaking with him afterward made the event feel less like extra credit and more like an actual step toward understanding what professional networking looks like.

## What Inspired Me

What inspired me most was that Elijah made the path into CS feel reachable. He did not present success as something that just happens because someone is naturally talented. He made it feel like the result of curiosity, consistency, communication, and being willing to put yourself into conversations.

This connected to my personal interests because I like building projects, but I sometimes think of projects as isolated work. This event reminded me that projects become more meaningful when I can talk about them, ask for feedback, and connect them to real people.

## Interaction

I talked to Elijah Johnson during the event by answering and asking questions, and I also spoke to him afterward. Later, I connected with him on LinkedIn, which made the experience feel more real because it did not just end when the bell rang.

He explained the importance of being proactive: participate, ask thoughtful questions, keep building, and follow up professionally. That was useful because it gave me a direct example of how to turn a school event into a real connection.

## Depth of Experience

This event changed my thinking because it made networking feel less intimidating. Before, I thought of networking as something formal that adults do. After this event, I saw that it can start with simply being curious, asking a good question, and continuing the conversation afterward.

It connects to my CS class because our class already emphasizes projects, collaboration, and reflection. The speaker event helped me see that those same habits matter outside the classroom too. For my future goals, it reinforced that I should keep building technical proof of work while also getting better at explaining what I build.

## Final Reflection

Overall, this event was meaningful because it made computer science feel more human. It was not just about code or grades. It was about direction, confidence, and learning how to put myself in rooms where I can grow.

**Biggest takeaway:** My biggest takeaway was that a strong CS path is built by asking better questions, building proof of work, and following up with people.
