# Using the Math Class Topic Check

Result: **98%**. Source trace: College Board assignment screenshot in `ref_for_ap_prep`. I used this after missing a random-number-range question in the broad MCQ work.

```java
public class MathClassDrill {
    public static int rollDie() {
        return (int) (Math.random() * 6) + 1;
    }

    public static int randomBetween(int low, int high) {
        return (int) (Math.random() * (high - low + 1)) + low;
    }

    public static double distanceFromOrigin(int x, int y) {
        return Math.sqrt(Math.pow(x, 2) + Math.pow(y, 2));
    }
}
```

## Reflection

The biggest improvement was remembering that `(int)(Math.random() * n)` starts at 0 and stops at `n - 1`. Any shifted range has to add the lower bound after the cast.
