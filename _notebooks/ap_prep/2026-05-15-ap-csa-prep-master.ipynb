# AP CSA Prep

This is the traceable overview of my AP Computer Science A preparation. The strategy was deliberately simple: do a couple of broad MCQ sets, identify weak topics from the misses, patch those weak spots with small targeted quizzes and syntax drills, then grind released FRQs because every FRQ forces the same tracing, loop, ArrayList, 2D array, and method-design habits that show up on MCQ.

The two largest hubs are the [MCQ composite notebook](https://magic005.github.io/student/ap-prep/mcq-overview/) and the [FRQ composite notebook](https://magic005.github.io/student/ap-prep/frq-overview/). I also kept a separate [Learning Java syntax drill notebook](https://magic005.github.io/student/ap-prep/learning-java-syntax-drills/) because that work was more about fluency than scoring.

## What I Was Optimizing For

The 2026 AP CSA target was a digital Bluebook exam with the familiar split between multiple choice and four free-response questions. I used the official [AP CSA exam page](https://apcentral.collegeboard.org/courses/ap-computer-science-a/exam), the [2026 AP exam dates page](https://apstudents.collegeboard.org/exam-dates), the [Course and Exam Description](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-a-course-and-exam-description.pdf), and the [Java Quick Reference](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-a-java-quick-reference.pdf) as the guardrails. For FRQ expectations, I worked from College Board released questions through the [past exam questions page](https://apcentral.collegeboard.org/courses/ap-computer-science-a/exam/past-exam-questions).

My personal emphasis ended up being:

- **Tracing before answering:** most MCQ misses came from rushing through state changes, loop bounds, or recursion.
- **Write the method contract first:** FRQs got cleaner when I wrote down what a method returns and when it mutates state.
- **Arrays and lists under pressure:** the broad MCQ sets showed that 1D arrays, ArrayLists, and 2D traversal were the highest value places to practice.
- **Use FRQs as MCQ training:** FRQ code made me slower but more accurate on the exact kinds of snippets MCQs ask me to trace.
