## Game Draft

```java
public int getScore() {
    int totalPoints = 0;

    if (levelOne.goalReached()) {
        totalPoints += levelOne.getPoints();
    };

    if (levelOne.goalReached() && levelTwo.goalReached()) {
        totalPoints += levelTwo.getPoints();
    };

    if (levelOne.goalReached() && levelTwo.goalReached() && levelThree.goalReached()) {
        totalPoints += levelThree.getPoints();
    };

    if (isBonus()) {
        totalPoints *= 3;
    };

    return totalPoints;
};


public int playManyTimes(int num) {
    int maxPoints = 0;
    for (int plays = 0; plays < num; plays++) {
        play();
        if (getScore() > maxPoints) {
            maxPoints = getScore();
        };
    };
    return maxPoints;
};
```

## ReviewAnalysis and Data Drafts

```java
// 3.

// a.

public double getAverageRating() {
    int totalRat = 0;
    for (int i = 0; i <= allReviews.length - 1; i++) {
        totalRat += allReviews.getRating(i);
    };
    return (double)(totalRat)/allReviews.length;
};

// b.

public ArrayList<String> collectComments() {
    ArrayList<String> comments = new ArrayList<>();

    for (int i = 0; i <= allReviews.length - 1; i++) {
        if (allReviews.getComment(i).contains("!")) {
            if (allReviews.getComment(i).endsWith("!") || allReviews.getComment(i).endsWith(".")) {
                comments.add(i + "-" + allReviews.getComment(i));
            }
            else {
                comments.add(i + "-" + allReviews.getComment(i) + ".");
            }
        }
    }
    return comments;
};




// 4.

// a.

public void repopulate() {
    for (int r = 0; r <= grid.length - 1; r++) {
        for (int c = 0; c <= grid[0].length - 1; c++) {
            boolean valid = false;
            int check = 0;
            while (valid == false) {    
                check = (int)(Math.random() * (MAX)) + 1;
                if (check % 10 == 0 && check % 100 != 0) {
                    grid[r][c] = check;
                    valid = true;
                }
            }
        }
    }
}


// b.

public int countIncreasingCols() {
    int count = 0;

    for (int r = 1; r <= grid.length - 1; r++) {
        for (int c = 0; c <= grid[0].length - 1; c++) {
            boolean increasing = false;
            if (grid[r][c] >= grid[r-1][c]) {
               increasing = true;
            }

        }
    }
    return count;
}
```

```java
public int countIncreasingColsRepair() {
    int count = 0;
    for (int c = 0; c < grid[0].length; c++) {
        boolean increasing = true;
        for (int r = 1; r < grid.length; r++) {
            if (grid[r][c] < grid[r - 1][c]) {
                increasing = false;
            }
        }
        if (increasing) {
            count++;
        }
    }
    return count;
}
```

## Reflection

The `Game` question helped with conditional accumulation. The `Data` question exposed a common 2D mistake: I was checking cells but not preserving the result for a whole column. The repair version keeps the boolean at the column level, which is a pattern I reused later in MCQ tracing.
