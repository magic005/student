# while Loops Topic Check

Result: **100%**. Source trace: College Board assignment screenshot in `ref_for_ap_prep`. I used this to tighten the loop mechanics that showed up in both MCQ sets: guard condition, update location, and whether the loop can print nothing.

```java
public class WhileLoopDrill {
    public static int countDigits(int n) {
        if (n == 0) {
            return 1;
        }
        int count = 0;
        while (n > 0) {
            count++;
            n /= 10;
        }
        return count;
    }

    public static int sumUntilLimit(int limit) {
        int value = 1;
        int sum = 0;
        while (sum + value <= limit) {
            sum += value;
            value++;
        }
        return sum;
    }
}
```

## Reflection

My main check was whether the variable changes before or after a condition matters. That is exactly how College Board hides off-by-one errors in short MCQ snippets.
