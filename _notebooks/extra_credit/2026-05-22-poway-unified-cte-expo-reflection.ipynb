## Before the Event

Before attending, I expected the CTE Expo to be a showcase of different classes and pathways. I thought it would mostly be tables, posters, and students explaining what their programs do.

I was curious about which booths would feel the most hands-on. Since I am interested in CS and engineering, I wanted to see which pathways connected classroom learning to real tools, real projects, and actual problem solving.

## What I Saw

I visited several booths and presentations, including business, plant/agriculture displays, photography/media work, animal science, and the Poway High School automotive group. I also gave feedback to people presenting, which made the event feel interactive rather than just something I walked through.

The most interesting thing I saw was the Poway High School automotive group. I liked that their class is not only theoretical. They work on actual cars, learn the engineering principles behind the systems, and also contribute to their school's robotics teams. That combination of mechanical work and robotics stood out to me.

## What I Learned

One specific thing I learned is that CTE classes can make engineering ideas much more concrete. Automotive work is full of systems thinking: mechanical parts, electrical behavior, diagnostics, safety, and teamwork all have to come together.

What surprised me was how connected the automotive pathway was to robotics. I usually separate cars, engineering, and robotics in my head, but seeing students connect hands-on vehicle work with robotics made me realize how much overlap there is between hardware, software, and problem solving.

## What Inspired Me

The Poway High automotive group inspired me the most because their work felt real. They were not just simulating engineering; they were touching real systems, fixing real things, and learning from machines that have consequences if something is misunderstood.

This connected to my personal interests because I like CS, but I also like when code connects to the physical world. Robotics, sensors, control systems, diagnostics, and automation all sit between software and hardware, and the automotive booth made that connection easier to picture.

## Interaction

I talked to student presenters and learned from the people running the booths. I asked questions, listened to their explanations, and gave feedback when I had reactions to their displays or projects.

The automotive students explained that they work directly with cars and use that experience to understand engineering principles. They also showed how their skills can support robotics, which made the booth feel especially relevant to CS.

## Connection to This CS Program and Outside CS Endeavors

The Expo also made me appreciate how far ahead this CS program is compared to a lot of normal school experiences. In this class, we are not only learning vocabulary or doing isolated assignments. We are building, publishing, reflecting, connecting projects together, and learning how to explain our work in a way that feels closer to an actual technical portfolio.

That connected directly to what I saw from the automotive group and their robotics involvement. Their pathway showed the hardware side of real engineering: physical systems, testing, iteration, and teamwork. This class gives me the software side of that same mindset, especially through project work, debugging, documentation, and deployment. Together, they made me think more about outside CS endeavors where code is not just on a screen, but part of something larger like robotics, automation, sensors, or tools that interact with real systems.

The biggest connection for me was that both the Expo and this CS program reward initiative. The students who seemed most impressive were the ones who had gone beyond minimum requirements, asked better questions, and had something real to show. That is the same direction I want to keep pushing in with CS outside of class.

## Depth of Experience

This event changed my thinking because it reminded me that technical learning is strongest when it has a physical or practical anchor. A classroom concept becomes easier to remember when I can see it in a car, a robot, a business pitch, a plant display, or a student project.

It connects to my CS class because CS is also about building systems that work in the real world. The Expo made me think more seriously about combining software with engineering, especially through robotics or hardware-connected projects. For my future goals, it pushed me to look for projects where code does something visible and useful.

## Final Reflection

Overall, the Poway Unified CTE Expo was meaningful because it showed how many different ways students can turn school into practical experience. I enjoyed seeing the variety of booths, but the automotive group especially gave me a clearer picture of engineering as something hands-on, collaborative, and connected to robotics.

**Biggest takeaway:** My biggest takeaway was that the best technical learning happens when classroom ideas become real systems you can build, test, fix, and explain.
