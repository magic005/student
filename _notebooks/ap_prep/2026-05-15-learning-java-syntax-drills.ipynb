## Student Class Basics

```java
public class Student {
    private String name;
    private double gpa;

    public Student(String n, double g) {
        name = n;
        gpa = g;
    }

    public String getName() {
        return name;
    }

    public double getGPA() {
        return gpa;
    }
}

Student test = new Student("Ansh", 4.0);
System.out.print(test.getGPA());
```

## Boolean Method and toString

```java
public class Student {
    private String name;
    private double gpa;

    public Student(String n, double g) {
        name = n;
        gpa = g;
    }

    public String getName() {
        return name;
    }

    public double getGPA() {
        return gpa;
    }

    public boolean isHonorRoll() {
        return gpa >= 3.5;
    }

    public String toString() {
        return name + " (GPA: " + gpa + ")";
    }
}

Student test = new Student("Ansh", 4.0);
System.out.print(test.toString());
```

## 2D Array Creation and Traversal

```java
int[][] arr = new int[2][3];

for (int r = 0; r < arr.length; r++) {
    for (int c = 0; c < arr[0].length; c++) {
        arr[r][c] = 7;
    }
}

int[][] table = {{1,2,3},{4,5,6},{7,8,9}};
int sum = 0;

for (int r = 0; r < table.length; r++) {
    for (int c = 0; c < table[0].length; c++) {
        sum += table[r][c];
    }
}

System.out.println(sum);
```

## Row-Specific Traversal

```java
int[][] data = {{1,2},{3,4},{5,6}};

for (int c = 0; c < data[1].length; c++) {
    System.out.println(data[1][c]);
}
```

## ArrayList Basics and Safe Removal

```java
import java.util.ArrayList;

ArrayList<String> fruits = new ArrayList<String>();
fruits.add("apple");
fruits.add("kiwi");
fruits.add("grape");

System.out.println(fruits.size());
System.out.println(fruits.get(1));

ArrayList<Integer> nums = new ArrayList<Integer>();
nums.add(10);
nums.add(20);
nums.add(30);
nums.add(20);
nums.add(40);

for (int i = nums.size() - 1; i >= 0; i--) {
    if (nums.get(i) == 20) {
        nums.remove(i);
    }
}

System.out.println(nums);
```

## Reflection

The most important syntax habit here was reverse traversal when removing from an ArrayList. The 2D array work also helped me stop mixing up `array.length` for rows and `array[0].length` for columns.
