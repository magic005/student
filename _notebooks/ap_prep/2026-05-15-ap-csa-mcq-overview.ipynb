```java
public class MCQRepairPatterns {
    public static int minOfThree(int a, int b, int c) {
        if (a <= b && a <= c) {
            return a;
        } else if (b <= a && b <= c) {
            return b;
        }
        return c;
    }

    public static int countNegatives(int[] values) {
        int count = 0;
        for (int value : values) {
            if (value < 0) {
                count++;
            }
        }
        return count;
    }

    public static int sum2D(int[][] grid) {
        int sum = 0;
        for (int r = 0; r < grid.length; r++) {
            for (int c = 0; c < grid[0].length; c++) {
                sum += grid[r][c];
            }
        }
        return sum;
    }
}
```
